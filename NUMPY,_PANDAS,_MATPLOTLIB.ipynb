```python
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns

bulan =['jan','feb','mar','apr','mei','juni','juli']
penjualan =[150, 200, 100, 250, 125, 299, 300]

plt.figure(figsize=(6,4))

sns.barplot (x=bulan, y=penjualan, palette='muted')

plt.plot(
    range(len(bulan)),
    penjualan,
    color='red',
    marker='o',
    linestyle='-',
    linewidth=2,
    label='garis tren'
)

plt.title('Penjualan per bulan garis tren', fontsize=12, fontweight='bold')
plt.xlabel('Bulan')
plt.ylabel('Jumlah')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import plotly.express as px
import pandas as pd

df = pd.DataFrame({
    'bulan':['jan','feb','mar','apr','mei'],
    'penjualan':[150, 200, 100, 250, 125]
})

fig = px.bar(df, x='bulan', y='penjualan', title='Penjualan per bulan')

fig.update_layout(width=700, height=400)

fig.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

url = "https://raw.githubusercontent.com/Alam-A99/Pandas_Latih/10ebb3af000e930a44989075733f495d5c2f4be5/penjualan.csv"
df = pd.read_csv(url)
df

df['total'] = df[['jan', 'feb', 'mar', 'apr']].sum(axis=1)

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))

df_sorted = df.sort_values('total', ascending=False)

sns.barplot(data=df_sorted, x='nama', y='total', palette='viridis')

plt.title('Total Penjualan Per Orang')
plt.xlabel('Nama Karyawan')
plt.ylabel('Total Penjualan')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
